```python
import pandas as pd
from pathlib import Path

PROCESSED = Path(
    r"C:\Users\Adity\OneDrive\Desktop\RESEARCH\Project Work\Inflation-Forecasting-Dashboard\data\processed"
)
```

```python
cpi_long = pd.read_csv(
    PROCESSED / "cpi_long_model.csv"
)

cpi_long["date"] = pd.to_datetime(
    cpi_long["date"],
    format="%d-%m-%Y"
)

cpi_long = (
    cpi_long
    .sort_values("date")
    .reset_index(drop=True)
)

print(
    "CPI Long:",
    cpi_long["date"].min(),
    "→",
    cpi_long["date"].max()
)

display(cpi_long.head())
display(cpi_long.tail())
```

```python
cpi_wpi = pd.read_csv(
    PROCESSED / "cpi_wpi_common_model.csv"
)

cpi_wpi["date"] = pd.to_datetime(
    cpi_wpi["date"],
    format="%d-%m-%Y"
)

cpi_wpi = (
    cpi_wpi
    .sort_values("date")
    .reset_index(drop=True)
)

print(
    "CPI + WPI:",
    cpi_wpi["date"].min(),
    "→",
    cpi_wpi["date"].max()
)

display(cpi_wpi.head())
display(cpi_wpi.tail())
```

```python
cpi_wpi_ppi = pd.read_csv(
    PROCESSED / "cpi_wpi_ppi_common_model.csv"
)

cpi_wpi_ppi["date"] = pd.to_datetime(
    cpi_wpi_ppi["date"],
    format="%d-%m-%Y"
)

cpi_wpi_ppi = (
    cpi_wpi_ppi
    .sort_values("date")
    .reset_index(drop=True)
)

print(
    "CPI + WPI + PPI:",
    cpi_wpi_ppi["date"].min(),
    "→",
    cpi_wpi_ppi["date"].max()
)

display(cpi_wpi_ppi.head())
display(cpi_wpi_ppi.tail())
```

```python
print("========== DATE VALIDATION ==========")

for name, df in {
    "CPI Long": cpi_long,
    "CPI + WPI": cpi_wpi,
    "CPI + WPI + PPI": cpi_wpi_ppi
}.items():

    expected = pd.date_range(
        start=df["date"].min(),
        end=df["date"].max(),
        freq="MS"
    )

    missing = expected.difference(df["date"])

    print(f"\n{name}")
    print("Start:", df["date"].min())
    print("End:", df["date"].max())
    print("Rows:", len(df))
    print("Duplicate dates:", df["date"].duplicated().sum())
    print("Missing calendar months:", len(missing))

print("\n=====================================")
```

```python
cpi_long.to_csv(
    PROCESSED / "cpi_long_model.csv",
    index=False
)

cpi_wpi.to_csv(
    PROCESSED / "cpi_wpi_common_model.csv",
    index=False
)

cpi_wpi_ppi.to_csv(
    PROCESSED / "cpi_wpi_ppi_common_model.csv",
    index=False
)

print("Corrected model datasets saved successfully.")
```
