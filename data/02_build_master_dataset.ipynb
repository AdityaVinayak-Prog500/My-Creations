```python
wpi_month_columns = wpi.columns[4:]

print("Number of monthly columns:", len(wpi_month_columns))
print("First 5:", wpi_month_columns[:5].tolist())
print("Last 5:", wpi_month_columns[-5:].tolist())
```

```python
wpi_long = wpi_core.melt(
    id_vars=[
        "Level",
        "Commodity Name",
        "Commodity Code",
        "Commodity Weight"
    ],
    value_vars=wpi_month_columns,
    var_name="month",
    value_name="index"
)

display(wpi_long.head(10))
```

```python
wpi_long["date"] = pd.to_datetime(
    wpi_long["month"],
    format="%b-%y"
)
wpi_long = wpi_long.sort_values(
    ["date", "Commodity Name"]
).reset_index(drop=True)
```

```python
display(
    wpi_long[
        ["date", "Commodity Name", "index"]
    ].head(20)
)
```

```python
wpi_long["series"] = wpi_long["Commodity Name"].map({
    "All Commodities": "wpi_all",
    "I PRIMARY ARTICLES": "wpi_primary",
    "II FUEL & POWER": "wpi_fuel",
    "III. MANUFACTURED PRODUCTS": "wpi_manufacturing"
})
print(wpi_long["series"].unique())
```

```python
wpi_master = wpi_long.pivot(
    index="date",
    columns="series",
    values="index"
).reset_index()
wpi_master.columns.name = None
```

```python
display(wpi_master.head())
display(wpi_master.tail())
```

```python
wpi_master = wpi_long.pivot(
    index="date",
    columns="series",
    values="index"
).reset_index()
wpi_master.columns.name = None
```

```python
expected_wpi_dates = pd.date_range(
    wpi_master["date"].min(),
    wpi_master["date"].max(),
    freq="MS"
)

missing_wpi_dates = expected_wpi_dates.difference(
    wpi_master["date"]
)

print("Expected months:", len(expected_wpi_dates))
print("Actual months:", len(wpi_master))
print("Missing months:", len(missing_wpi_dates))
print(missing_wpi_dates)
```

```python
for column in [
    "wpi_all",
    "wpi_primary",
    "wpi_fuel",
    "wpi_manufacturing"
]:
    wpi_master[f"{column}_yoy"] = (
        wpi_master[column]
        .pct_change(12)
        * 100
    )
```
